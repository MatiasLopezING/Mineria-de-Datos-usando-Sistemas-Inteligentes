```python
# 2 intervalos de igual ancho entre min y max
df_estrellas["lum_intervalo_rango"], bins = pd.cut(df_estrellas["Luminosidad"], bins=2, retbins=True)

# límites numéricos de corte
print("Cortes:", bins)

# intervalos creados
print("Intervalos:", df_estrellas["lum_intervalo_rango"].cat.categories)

# cantidad de ejemplos por intervalo
print(df_estrellas["lum_intervalo_rango"].value_counts().sort_index())
```


### C. Correlación lineal
Calcule la correlación lineal entre los atributos **Luminosidad** y **Temperatura**. Indique la intensidad de la correlación (no hay correlación / débil / fuerte) y el tipo (positiva / negativa).

| Atributo | Valor |
| :--- | :--- |
| **Intensidad** | |
| **Tipo** | |


```python
# Calcular correlación de Pearson entre Luminosidad y Temperatura
correlacion = df_estrellas["Luminosidad"].corr(df_estrellas["Temperatura"])

print(f"Correlación Lineal: {correlacion:.4f}")

# Determinar intensidad y tipo
if abs(correlacion) < 0.3:
    intensidad = "No hay correlación"
elif abs(correlacion) < 0.7:
    intensidad = "Débil"
else:
    intensidad = "Fuerte"

tipo = "Positiva" if correlacion > 0 else "Negativa"

print(f"\nIntensidad: {intensidad}")
print(f"Tipo: {tipo}")
```

```python

# Diagrama de dispersión con línea de tendencia
plt.figure(figsize=(10, 6))
sns.regplot(data=df_estrellas, x="Luminosidad", y="Temperatura")
plt.title(f"Correlación entre Luminosidad y Temperatura\n(r = {correlacion:.4f})", fontsize=14)
plt.xlabel("Luminosidad", fontsize=12)
plt.ylabel("Temperatura", fontsize=12)
plt.grid(True, alpha=0.3)
plt.show()
```


### D. Diagrama de Caja de Tukey
Dibuje un Diagrama de Caja de Tukey de la variable **Luminosidad** e inclúyalo en la respuesta. Indique también los valores del cuadro:

| Parámetro | Valor |
| :--- | :--- |
| **Mediana** | |
| **Q1** | |
| **Q3** | |
| **RI (Rango Intercuartil)** | |
| **Bigote superior** | |
| **Bigote inferior** | |
| **Intervalos de valores atípicos leves** | |
| **Valores atípicos leves** | |
| **Intervalos de valores atípicos extremos** | |
| **Valores atípicos extremos** | |

```python
# Calcular parámetros del Diagrama de Caja de Tukey
lum = df_estrellas["Luminosidad"]

# Cuartiles
Q1 = lum.quantile(0.25)
Q2 = lum.quantile(0.50)  # Mediana
Q3 = lum.quantile(0.75)

# Rango Intercuartil
RI = Q3 - Q1

# Bigotes
bigote_inferior = max(lum.min(), Q1 - 1.5 * RI)
bigote_superior_limite = Q3 + 1.5 * RI
bigote_superior = lum[lum <= bigote_superior_limite].max()

# Valores atípicos
atipicos_leves_inf = Q1 - 1.5 * RI
atipicos_leves_sup = Q3 + 1.5 * RI

atipicos_extremos_inf = Q1 - 3 * RI
atipicos_extremos_sup = Q3 + 3 * RI

valores_atipicos_leves = lum[(lum < Q1 - 1.5*RI) | (lum > Q3 + 1.5*RI)]

valores_atipicos_extremos = lum[(lum < Q1 - 3 * RI) | (lum > Q3 + 3 * RI)]

print("=== DIAGRAMA DE CAJA DE TUKEY - LUMINOSIDAD ===\n")
print(f"Mediana (Q2): {Q2:.4f}")
print(f"Q1 (Cuartil 1): {Q1:.4f}")
print(f"Q3 (Cuartil 3): {Q3:.4f}")
print(f"RI (Rango Intercuartil): {RI:.4f}")
print(f"\nBigote inferior: {bigote_inferior}")
print(f"Bigote superior: {bigote_superior}")
print(f"\nIntervalo valores atípicos leves: [{atipicos_leves_inf:.4f}, {atipicos_leves_sup:.4f}]")
print(f"Valores atípicos leves: {list(valores_atipicos_leves.values)}")
print(
    f"\nIntervalo valores atípicos extremos: [{atipicos_extremos_inf:.4f}, {atipicos_extremos_sup:.4f}]"
)

print(f"Valores atípicos extremos: {list(valores_atipicos_extremos.values)}")
```
